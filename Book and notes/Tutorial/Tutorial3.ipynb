### 1. Create a  Car  class with attributes  make ,  model , and  year . Add a method to display car details.

```python
class Car:
    def __init__(self,make,model,year):
        self.make=make
        self.model=model
        self.year=year

    def car_details(self):
        print("Car details")
        print(f"Make: {self.make}" )
        print(f"Model: {self.model}" )
        print(f"Year: {self.year}" )
        

car1 = Car("BMW",1200, 2018)
car1.car_details()
```

```python
# Create a Circle class with attribute radius . Use encapsulation to protect radius and add getter and setter methods. Add a method to calculate the area.
import math
class Circle:
    def __init__(self,radius):
        self.__radius = radius

    def get_radius(self):
        return self.__radius

    def set_radius(self):
        if radius>0:
            self.radius=radius
        else:
            print("the radius must be positive.")

    def calculate_area(self):
        return math.pi * (self.__radius ** 2)

c1 = Circle(5)
c1.calculate_area()
```

```python
# Create a BankAccount class. Add methods for deposit, withdraw, and checking balance. Ensure the balance cannot be accessed directly.

class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance # private attribute
    def deposit(self, amount):
        self.__balance += amount
    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Insufficient funds")
    def get_balance(self):
        return self.__balance
        
account = BankAccount("Alice", 1000)
account.deposit(500)
account.withdraw(200)
print(account.get_balance()) # 1300
```

```python
# Create a Student class with attributes name , roll_number , and marks . Add methods to display student info and calculate average marks.

class Student:
    def __init__(self, name, roll_number, marks):
        self.name = name
        self.roll_number = roll_number
        self.marks = marks   # marks should be a list

    # Method to display student information
    def display_info(self):
        print(f"Name        : {self.name}")
        print(f"Roll Number : {self.roll_number}")
        print(f"Marks       : {self.marks}")

    # Method to calculate average marks
    def calculate_average(self):
        if len(self.marks) == 0:
            return 0
        return sum(self.marks) / len(self.marks)


# Example usage
s1 = Student("Prajwal", 101, [85, 90, 78, 92])

s1.display_info()
print("Average Marks:", s1.calculate_average())
```
