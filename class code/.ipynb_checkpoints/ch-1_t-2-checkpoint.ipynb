1. Factorial Using while Loop

```python
f = 1
n = 5

while n >= 1:
    f = f * n
    n = n - 1

print(f)
```

2. Pattern Printing

```python
for i in range(1, 6):
    for j in range(1, i+1):
        print(i, end=" ")
    print()
```

3. Sum of Digits

```python
n = 12345
s = 0

while n != 0:
    rem = n % 10
    s = s + rem
    n = n // 10

print(s)
```

```python
def add_numbers(*args):
    print(args)
add_numbers(1, 2, 3, 4)

```

```python
def total_sum(*args):
    s = 0
    for num in args:
        s += num
    return s

print(total_sum(10, 20, 30))
print(total_sum(1, 2, 3, 4, 5))
```

```python
def student_info(**kwargs):
    print(kwargs)

student_info(name="Prajwal", age=21, course="Data Science")
```

```python
def student_info(**kwargs):
    for key, value in kwargs.items():
        print(key, ":", value)

student_info(name="Prajwal", age=21)
```

```python
def show_data(*args, **kwargs):
    print("Args:", args)
    print("Kwargs:", kwargs)

show_data(10, 20, name="Prajwal", score=95)
```

```python
def describe_data(*values, **info):
    print("Values:", values)
    print("Metadata:",info)
    for k, v in info.items():
        print(k, "=", v)

describe_data(10, 20, 30, source="sensor", unit="cm")
```
