## Iterator


```python
print("hello world")
alist = [1,2,3,4]
it = iter(alist)
# print(it)
# print(type(it))
while True:
    try:
        num = next(it)
        print(num)
    except StopIteration:
        break
```

## Decorator

```python
def myDecorator(func):
    def wrapper():
        print("Before")
        func()
        print("After")
    return wrapper

@myDecorator
def sayHello():
    print("Hello")

sayHello()
```

```python
import time 
def timer(func):
    def wrapper(args,*kwargs):
        start = time.time()
        result = func(args,*kwargs)
        end = time.time()
        print(f"Elapsed time= {end-start} seconds")
        return result
    return wrapper

@timer
def fact(n):
    f =1
    for i in range(1,n+1):
        f = f*i
    return f
fact(100)
```

 ## Generator

```python
def squares(x):
    for i in range(x):
        yield i*i
gen = squares(100)
for i in gen :
    print(i)
# print(squares(5))
# print(type(squares(5)))
```
