1. Exception Handling

```python
try:
    n1 = int(input("Enter a number"))
    n2 = int(input("Enter another number"))
    print(n1 / n2)

except ZeroDivisionError:
    print("Cannot divide by zero")

finally:
    print("This always runs")
```

2. Counter

```python
from collections import Counter

aList = ["Ram", "Shyam", "Ram"]
c = Counter(aList)
print(c)
```
