3. Plotting a Graph (Data Visualization)

```python
import numpy as np
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [10, 20, 25, 30]

plt.plot(x, y, label="X vs Y")
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.title("X vs y")
plt.legend()
plt.savefig("plot.png")
```

4. Factorial Using Loop

```python
f = 1
for i in range(1, 101):
    f = f * i

print(f)
print(len(str(f)))
```
