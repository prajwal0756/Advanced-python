Feature Transformation and scaling

```python
#log transformation
import numpy as np
import pandas as pd 
x= pd.Series([0,1,10,100,1000])#1d obj
x_log = np.log1p(x)
print(x_log)
```

```python
# Square root transformation

import numpy as np
import pandas as pd 
x= pd.Series([0,1,10,100,1000])#1d obj
x_sqrt = np.sqrt(x)

print(x_sqrt)
```

```python
import matplotlib.pyplot as plt
plt.plot(x,label="x")
plt.plot(x_log,label="x_log")
plt.legend()
```

Power Transformation


```python
from sklearn.preprocessing import PowerTransformer
pt= PowerTransformer(method='yeo-johnson')

x_transformed = pt.fit_transform(x.values.reshape(-1, 1))
print (x_transformed)
```

Normalization

```python
# minmax normalization

from sklearn.preprocessing import MinMaxScaler
x=np.array([[100],[200],[800],[900],[1000]])
scaler=MinMaxScaler()
x_transforemed=scaler.fit_transform(x)
print(x_transforemed)

#error check it
```

```python
# z-score normalization


from sklearn.preprocessing import StandardScaler
x=np.array([[100],[200],[800],[900],[1000]])
scaler=StandardScaler()
x_transforemed=scaler.fit_transform(x)
print(x_transforemed)
```

```python
import numpy as np
# Example data
data = np.array([200, 300, 400, 600, 1000])
# Find j
max_val = np-max (np.abs (data))
j = len(str(int (max_val)))
# Apply decimal scaling
scaled_data = data / (10 ** j)
print("Original data:", data)
print("Scaled data:", scaled_data)
```

Feature Encoding

```python
# Label encoder


from sklearn preprocessing import LabelEncoder
le = LabelEncoder ()
cities = ['Kathmandu', 'Pokhara', 'Biratnagar']
encoded = le.fit_transform(cities)
print (encoded)
```

```python
# One Hot Encoding

import pandas as pd
df = pd.DataFrame({'source': ['river', 'lake', 'well'] })
df_encoded = pd.get_dummies(df, columns=['source' ])
print(df_encoded)
```

```python
# Ordinal Encoding

from sklearn preprocessing import OrdinalEncoder
encoder = OrdinalEncoder(categories=[['poor', 'fair', 'good']])
quality = [['poor'], ['good'], ['fair']]
encoded = encoder.fit_transform(quality)
print (encoded)
```
