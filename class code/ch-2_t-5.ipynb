```python
import pandas as pd

import random 
names =["Ram","shyam","hari","krishna","sita"]
final_list=[]
for i in ( range(1,10000001)):
    random_name = names[random.randint(0,4)]
    final_list.append({
        "Eid":i,
        "Name":random_name,
        "salary":100000*random.random()
    })

df= pd.DataFrame(final_list)
df.to_csv("big-data.csv", index= False)
```

```python
 first_chunk =True
output_file ='processed_data.csv'
for chunk in pd.read_csv('big-data.csv',chunksize =100000):
    chunk["Gender"]= chunk["Name"].apply(lambda x: "Female" if x=="sita" else "Male")
    chunk.to_csv(
        output_file,
        index =False,
        mode="w" if first_chunk else "a",
        header =first_chunk
    )
    first_chunk=False
```

```python
def read_lines (filename):
    with open(filename) as f:
    for line in f:
    yield line

# created only when needed
for line in read_lines ("example. txt"):
    print (line)

```
