```python
# Number of words in a text file
with open(file='try.txt', mode='r') as f:
    content = f.read()
    print(len(content.split()))
```

```python
# frequency of words in a text file
aDict = {}
with open(file='try.txt', mode='r') as f:
    content = f.read().lower()
    words = content.split()
    for word in words:
        if word not in aDict:
            aDict[word] = 1
        else:
            aDict[word] += 1
            
print(aDict)
```

```python
with open(file="new.txt", mode = "w") as f:
    input_text = input("Enter some text")
    f.write(input_text)
```

```python
with open(file="new.txt", mode = "w") as f:
    while(True):
        input_text = input("Enter some text. Press Q or q to quit")
        if input_text == 'Q' or input_text == 'q':
            break
        f.writelines(input_text)
        f.write('\n')
```

```python
import csv
data = [
    {'Name': 'John', 'Address': 'Biratnagar', 'Age': '22'},
    {'Name': 'Jane', 'Address': 'Itahari', 'Age': '21'},
    {'Name': 'Hari', 'Address': 'Dharan', 'Age': '23'}
]
with open(file="neww.csv",mode='w',newline="") as f:
    writer = csv.DictWriter(f, fieldnames=['Name', 'Address', 'Age'])
    writer.writeheader()
    writer.writerows(data)
```
